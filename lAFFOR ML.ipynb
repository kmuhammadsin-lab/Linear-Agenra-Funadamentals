```python
import numpy as np

# Create a 2x2 matrix
A = np.array([
    [4, 7],
    [2, 6]
])

# Calculate determinant
det = np.linalg.det(A)

# Calculate inverse
inverse = np.linalg.inv(A)

print("Original Matrix:")
print(A)

print("\nDeterminant:")
print(det)

print("\nInverse Matrix:")
print(inverse)
```

properties of matrices[link text](https://)

```python
import numpy as np

# Create matrices
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

C = np.array([
    [2, 0],
    [1, 2]
])

# 1. Associative Property
left = (A @ B) @ C
right = A @ (B @ C)

print("Associative Property:")
print(np.allclose(left, right))

# 2. Distributive Property
left = A @ (B + C)
right = (A @ B) + (A @ C)

print("\nDistributive Property:")
print(np.allclose(left, right))

# 3. Identity Property
I = np.eye(2)

result = A @ I

print("\nIdentity Property:")
print(np.allclose(result, A))

# 4. Non-Commutative Property
AB = A @ B
BA = B @ A

print("\nAB:")
print(AB)

print("\nBA:")
print(BA)

print("\nAre AB and BA equal?")
print(np.allclose(AB, BA))
```

Create a Block Diagonal Matrix Using NumPy

```python
import numpy as np

# Create first matrix
A = np.array([
    [1, 2],
    [3, 4]
])

# Create second matrix
B = np.array([
    [5, 6],
    [7, 8]
])

# Create block diagonal matrix
D = np.block([
    [A, np.zeros((2, 2))],
    [np.zeros((2, 2)), B]
])

print("Block Diagonal Matrix:")
print(D)
```
